```python
def f(num):
    
    """
    Calculates the function f(x) = x^3 - x^2 - 1

    Parameters

    x : float
        Input value

    Returns

    float
        Value of f(x)
    """
    
    return num**3-num**2-1
```

```python
print(f(1))
```

```python
def df(num):
    
    """
    Calculates the derivative of f(x)

    The derivative is:
    df/dx = 3x^2 - 2x

    Parameters
  
    x : float
        Input value

    Returns

    float
        Value of df/dx
    """

    return (3*num**2)-(2*num)
```

```python
print(df(1))
```

```python
def newton(f, df, x0, epsilon=1e-6, max_iter=30):
    """
    Uses Newton's Method to find a root of a function.

    Parameters

    f : function whose root is being found

    df : function - derivative of f

    x0 : float - initial guess for the root

    epsilon : float, optional
        desired accuracy - default is 1e-6

    max_iter : int, optional
        maximum number of iterations - default is 30

    Returns

    float or None
        Approximate root if successful, otherwise None
    """

    x = x0

    for i in range(max_iter):

        if abs(f(x)) < epsilon:
            print(f"Found root in {i} iterations")
            return x

        x = x - f(x) / df(x)

    print("Iteration failed")
    return None
```

```python
# example 1
root1 = newton(f, df, x0=1.5)
print("Root:", root1)

print()

# example 2
root2 = newton(f, df, x0=2.0)
print("Root:", root2)

print()

# using smaller epsilon
root3 = newton(f, df, x0=1.5, epsilon=1e-8)
print("Root with epsilon=1e-8:", root3)
```
